# UDA-Hub Demo

This notebook demonstrates the UDA-Hub LangGraph workflow with:
1. A resolved ticket (onboarding question matched to KB article)
2. An escalated ticket (complex billing dispute)
3. Multi-turn conversation with context preservation
4. Customer-specific memory recall

```python
import json
from dotenv import load_dotenv
from agentic.workflow import compile_graph

load_dotenv()
graph = compile_graph()
print("Graph compiled successfully")
```

## Demo 1: Resolved Ticket (Onboarding)

```python
state = {
    "ticket_id": "demo-resolved-001",
    "ticket_text": "How do I get started with CultPass and reserve my first event?",
    "user_email": "newuser@example.com",
    "customer_id": "newuser@example.com",
    "classification": {},
    "resolution": {},
    "escalation": {},
    "tool_results": [],
    "memory_context": [],
    "agent_trace": [],
    "messages": [],
}
config = {"configurable": {"thread_id": "demo-resolved-001"}}

result = graph.invoke(input=state, config=config)

print("=== RESOLVED TICKET ===")
print(f"Category: {result['classification'].get('category')}")
print(f"Routing: {result['classification'].get('routing_label')}")
print(f"Resolution: {result['resolution'].get('status')} (confidence={result['resolution'].get('confidence', 0):.2f})")
print(f"Article: {result['resolution'].get('article_title')}")
print(f"Response: {result['resolution'].get('response', '')[:200]}")
print(f"Agent trace: {' -> '.join(result['agent_trace'])}")
```

## Demo 4: Tool Usage Inspection

```python
state = {
    "ticket_id": "demo-tools-001",
    "ticket_text": "I need to update my payment method for my CultPass subscription. Can you look up my account details?",
    "user_email": "billing@example.com",
    "customer_id": "billing@example.com",
    "classification": {},
    "resolution": {},
    "escalation": {},
    "tool_results": [],
    "memory_context": [],
    "agent_trace": [],
    "messages": [],
}
config = {"configurable": {"thread_id": "demo-tools-001"}}

result = graph.invoke(input=state, config=config)

print("=== TOOL USAGE ===")
print(f"Category: {result['classification'].get('category')}")
print(f"Routing: {result['classification'].get('routing_label')}")
print(f"Resolution: {result['resolution'].get('status')} (confidence={result['resolution'].get('confidence', 0):.2f})")
print(f"Agent trace: {' -> '.join(result['agent_trace'])}")
print(f"\nTool results ({len(result['tool_results'])} calls):")
for i, tr in enumerate(result['tool_results']):
    tool_name = tr.get('tool', 'unknown')
    if 'matches_found' in tr:
        print(f"  {i+1}. {tool_name}: {tr['matches_found']} matches found")
    elif 'result' in tr:
        r = tr['result']
        if isinstance(r, dict) and 'error' in r:
            print(f"  {i+1}. {tool_name}: {r['error']}")
        else:
            print(f"  {i+1}. {tool_name}: {r}")
    else:
        print(f"  {i+1}. {tool_name}: {tr}")
```
